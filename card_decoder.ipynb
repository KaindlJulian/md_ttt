# Card Decoder

You are assigned 1 Hidden Target Monster Card. Your objective is to guess the exact card (or match its absolute identity) within the following constraints:
 
To successfully "decode" the card, your guesses must align across six distinct structural properties:
- Frame: Effect, Fusion, Synchro, Xyz, Link, Ritual, Normal and combinations with Pendulum
- Attribute: DARK, LIGHT, EARTH, FIRE, WATER, WIND, DIVINE
- Level / Rank / Link Rating: The discrete scale value (integers)
- Type: Warrior, Spellcaster, Dragon, Fiend, Cyberse, etc
- ATK: The attack stat line (Discrete values)
- DEF: The defense stat line (Discrete values, or none for Link Monsters)

Guesses (3) reveal the correct/incorrect properties of the given card

Hints (3) reveal 1 correct property

# Data Exploration and Mode Calculation

looking at which properties and their distinct values

need the mode of each property individually, because challenge allows marginal elimination.
if a card exists that best represents the marginal modes we take it as a first guess

```python
import json
import os
import pandas as pd
from tqdm.notebook import tqdm

JSON_PATH = "cardinfo_full.json"  # https://www.ygoprodeck.com/api-guide/ , v7 ygoprodeck api json dump

with open(JSON_PATH, 'r') as f:
    cards = json.load(f)

df = pd.DataFrame(cards['data'])

# filter out monsters only
df = df[df['type'].str.contains('Monster')]

# rename race to type
df = df.drop(columns=['type'])
df = df.rename(columns={'race': 'type'})  


def get_level(row):
    if 'level' in row and not pd.isna(row['level']):
        return row['level']
    elif 'rank' in row and not pd.isna(row['rank']):
        return row['rank']
    elif 'linkval' in row and not pd.isna(row['linkval']):
        return row['linkval']
    else:
        return None
    
# level/rank/link is one property in the challenge, so we can combine them into one new column
df['levelRankLink'] = df.apply(get_level, axis=1)

for col in ['frameType', 'attribute', 'type', 'levelRankLink', 'atk', 'def']:
    unique = df[col].unique()
    print(f"Distinct values in {col} (count: {len(unique)}): {unique}")
```
